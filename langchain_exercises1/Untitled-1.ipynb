```python
ct =0
def recursion(ct):
    if ct == 20: #base case

        return 0
    else:
        print(f"Recursion num  {ct}")
        recursion(ct+1)
```

```python
recursion(ct)
```

```python
def factorial_using_recursion(n:int) -> int:
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return n * factorial_using_recursion(n-1)
```

```python
fact = factorial_using_recursion(6)
print(fact)
```

```python
factorial_using_recursion(2)
```
