```python
from langchain_core.messages import AIMessage

messages = [
    (
        "system",
        "You are a helpful assistant that translates English to Hindi and Kannada. Translate the user sentence.",
    ),
    ("human", "I love programming."),
]
ai_msg = model.invoke(messages)
ai_msg
```

## Messages

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(
        content="You are a bad ass assistant! Your name is Bob. \
            You will revert users in cool slangs and explict words"
    ),
    HumanMessage(
        content="What is your name?"
    )
]

# Instantiate a chat model and invoke it with the messages
# model = ...

    
revert = model.invoke(messages)
print(revert.content)
```

```python
messages
```

```python
for chunk in model.stream(messages):
    print(chunk.content)
```
